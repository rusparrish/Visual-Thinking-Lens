
# VCLI-G Methods (Geometry-Coupled Visual Cognitive Load Index)

VCLI-G is a research-grade scoring system designed to measure how much perceptual and cognitive work an image demands — not how “pretty” it is, not how similar it is to training data, and not how well it matches a text prompt.
It couples perceptual arrest (how hard an image is to resolve or “let go of”) with geometric signals extracted directly from the composition.

It’s meant to answer a single, difficult question:

How much does this image make me think?

The system is built around four measurable geometric signals:

G1 — Centroid Wander (attention instability): how much the visual center shifts as you scan the image.

G2 — Void Topology (figure/ground ambiguity): how space and absence are organized, and whether they resist resolution.

G3 — Curvature Torque (formal tension): how much structural pressure or opposition is present in directional flow.

G4 — Occlusion Entropy (proxy) (depth/ambiguity): how near/far relationships and overlaps create perceptual uncertainty.

What this is not

❌ A style or aesthetics score — it does not reward “beautiful” images.

❌ A similarity or CLIP-like model — it does not care how close an image is to its text prompt.

❌ A quality ranking — a low score can mean “intentionally simple” or “fully resolved,” not “bad.”

❌ A finished metric — this is a scaffold, designed to be clear, auditable, and extendable.

❌ It is not a composition or aesthetics score. It does not judge balance, color harmony, polish, or “good design.”

Score	Cognitive Load	Viewer Behavior	Meaning

VCLI-G returns a composite score on a 0.0 – 5.0 scale:

0.0 – 1.0 → Low cognitive load – The image resolves quickly. It’s stable, obvious, or visually quiet.

1.0 – 3.0 → Moderate load – There’s some structural complexity, but resolution is straightforward.

3.0 – 4.0 → High load – The image resists closure. Tension, contradiction, or ambiguity hold your attention.

4.0 – 5.0 → Very high load – The image “won’t let go.” It’s perceptually sticky, demands re-entry, or unfolds across multiple viewings.

Treat VCLI-G as a delay index — a measure of how long an image resists resolution, not how “good” it is.

Compare scores to understand viewer effort or attention gravity across images or runs.

Use it alongside structural tools to reveal where and why perceptual strain occurs — but don’t confuse high scores with aesthetic merit

⚠️ **Important:**
- A high VCLI-G score does not mean "better." It means "more cognitively demanding."
- A high SCI score does not mean "better." It means "more organized."
- In some artistic or design contexts, that’s desirable. In others, a low-load image may be the right result.
- In some contexts, low VCLI-G + high SCI is ideal (branding, UI).
- In others, high VCLI-G + low SCI is valuable (experimental art, emergent process).
- The scores describe what's happening, not whether it's good.

Why this matters for AI

Modern reward models and diffusion pipelines are tuned to favor images that resolve quickly — centered, balanced, clear, and “understandable.”
That’s useful for general purpose generation, but it suppresses the very qualities artists often care about: tension, delay, contradiction, recursive looking.

VCLI-G offers a way to quantify and reintroduce those qualities into evaluation loops.
Used alongside CLIP scores, aesthetic ratings, or model logits, it helps researchers and artists:

Detect when a model is playing it safe visually.

Explore regions of the latent space that produce consequential work.

Benchmark changes in perceptual engagement across model versions or prompt strategies.

## Two Metrics, One System

VCLI-G is now paired with **SCI (Structural Coherence Index)** to form a 2D perceptual space:

**VCLI-G (0.0 – 5.0)**: How much cognitive work does the image demand?
**SCI (0.0 – 5.0)**: How organized is that work?

They're independent axes that together describe different kinds of visual complexity:

**High VCLI-G, High SCI** → Earned tension (Cézanne, deliberate ambiguity)
**High VCLI-G, Low SCI** → Chaotic complexity (glitch, noise, productive accidents)
**Low VCLI-G, High SCI** → Resolved clarity (Vermeer, intentional simplicity)
**Low VCLI-G, Low SCI** → Default simplicity (gradient + centered object)

⚠️ **Neither axis is a quality score.** Low SCI doesn't mean "bad" — it means emergent/process-driven (like Pollock). High SCI doesn't mean "good" — it means systematic/composed. Context determines which is appropriate.

## SCI Score Bands

**0.0 – 1.5** → Entropic — High local variance, scattered edges, inconsistent detail. Complexity is emergent or random.

**1.5 – 3.0** → Mixed — Some organizational structure, but irregular. Parts may be coherent while others drift.

**3.0 – 4.0** → Coherent — Clear structural logic. Edges align, detail is consistent, rhythm is present.

**4.0 – 5.0** → Highly organized — Systematic, deliberate decisions visible throughout. Strong internal consistency.

**SCI measures organization, not intention.** An algorithmic pattern can score high SCI. A deliberate "messy" sketch can score low SCI. Both are valid depending on context.

## Why SCI Matters Alongside VCLI-G

VCLI-G alone can't distinguish:
- Intentional ambiguity from incoherent noise
- Earned tension from accidental chaos  
- Deliberate simplicity from lazy defaults

**SCI provides context.** It reveals whether high cognitive load comes from:
- **Organized friction** (high SCI) → deliberate formal tension, layered decisions
- **Entropic friction** (low SCI) → emergent complexity, process-driven, experimental

This lets you:
- Filter AI outputs for "playing it safe" (low VCLI-G, high SCI)
- Identify consequential work (high VCLI-G, high SCI)
- Detect when models collapse into noise (high VCLI-G, low SCI)
- Find productive accidents (low SCI but interesting)

**Use case examples:**
- **UI/UX design** → Filter for low VCLI-G, high SCI (clear, organized)
- **Gallery curation** → Hunt for high VCLI-G, high SCI (earned complexity)
- **Experimental work** → Embrace high VCLI-G, low SCI (productive chaos)
- **Model diagnostics** → Track when outputs drift into low SCI (overfitting to texture)

## Understanding the Three Views: CANON, APE, and RCE

This notebook generates three complementary views of the VCLI-G × SCI space:

**CANON (Geometric Base)** - The raw measurement layer. Deterministic and reproducible. Use this for absolute comparisons and quantitative reporting. Answers: "How much geometric sophistication exists?"

**APE (Anisotropic Power Equalization)** - Perceptual normalization via differential power-law scaling (γ = 1.2 for SCI, 2.0 for VCLI-G). Compresses the dense statistical center while preserving extreme outliers, converting geometric distance into perceptual distance. Answers: "Is this sophistication robust, or just clustering noise?"

**RCE (Residual Contrast Expansion)** - Local variance amplification within SCI coherence bands (k ≈ 1.6). Reveals strategic differences *within* sophistication tiers - whether high complexity comes from sustained tension (Cézanne) or resolved restraint (Corot). Answers: "What's the artistic strategy at this level of coherence?"

**Important:** APE and RCE are distribution-dependent. They normalize relative to the current batch's statistics, so adding/removing images will shift the perceptual field. This is intentional - it models how human perception recalibrates "complex" based on context. For reproducible comparisons, always use CANON scores. For interpretive visualization and cross-modal comparison (AI vs. human), APE/RCE reveal latent structure that raw scores compress.

**Quick guide:** Start with CANON to establish tiers. Use APE to confirm robust separation. Use RCE to distinguish strategies within masters.

```python

def load_image(path):
    img = cv2.imread(path, cv2.IMREAD_COLOR)
    if img is None:
        raise FileNotFoundError(f"Could not read image at {path}")
    return img

def to_gray(img):
    if img.ndim == 3:
        g = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        g = img.copy()
    return g.astype(np.float32) / 255.0

def saliency_map(gray):
    gx = filters.sobel_h(gray)
    gy = filters.sobel_v(gray)
    return np.hypot(gx, gy)

def saliency_pyramid(gray, sigmas=(2,4,8,16)):
    maps = []
    for s in sigmas:
        g = gaussian_filter(gray, sigma=s)
        maps.append(saliency_map(g))
    return maps
```

```python

def weighted_centroid(smap):
    sm = smap + 1e-8
    h, w = sm.shape
    y, x = np.mgrid[0:h, 0:w]
    m = sm.sum()
    cx = (sm * x).sum() / m
    cy = (sm * y).sum() / m
    return np.array([cx, cy])

def path_length(points):
    if len(points) < 2:
        return 0.0
    diffs = np.diff(points, axis=0)
    return np.linalg.norm(diffs, axis=1).sum()

def polyline_turn_curvature(points):
    """
    Mean absolute turning angle (radians) per segment.
    Stable for small N.
    """
    if len(points) < 3:
        return 0.0
    pts = np.asarray(points, dtype=float)
    v = np.diff(pts, axis=0)               # segment vectors
    # remove near-zero segments to avoid noise
    lens = np.linalg.norm(v, axis=1) + 1e-8
    v = v[lens > 1e-6]
    if len(v) < 2:
        return 0.0
    # normalize and compute turning angles
    v = v / (np.linalg.norm(v, axis=1, keepdims=True) + 1e-8)
    dots = np.clip((v[:-1] * v[1:]).sum(axis=1), -1.0, 1.0)
    angs = np.arccos(dots)                 # [0..pi]
    return float(np.mean(np.abs(angs)))

def G1_centroid_wander(gray, sigmas=(1.5, 3, 6, 12, 24)):
    """
    Multiscale centroids + path metrics.
    - More scales for better shape.
    - K from discrete turning angles (robust at low N).
    """
    pyr = saliency_pyramid(gray, sigmas=sigmas)
    pts = np.array([weighted_centroid(p) for p in pyr], dtype=float)

    # path length (pixels)
    diffs = np.diff(pts, axis=0)
    L = float(np.linalg.norm(diffs, axis=1).sum()) if len(pts) > 1 else 0.0

    # curvature (radians/turn)
    K = polyline_turn_curvature(pts)
    return {"points": pts, "path_len": L, "path_curv": K}
```

```python
# --- Void Topology v2 (clean) ---
# Brightness-thresholded voids + robust metrics (no wild Euler spikes)

import numpy as np
import cv2
from skimage.measure import label, regionprops, euler_number
from skimage.morphology import (
    remove_small_holes, remove_small_objects, binary_opening, binary_closing, disk
)
from scipy.ndimage import distance_transform_edt

def _edges_for_viz(gray):
    g8  = (gray * 255).astype(np.uint8)
    med = np.median(g8)
    lo  = int(max(0, (1.0 - 0.66) * med))
    hi  = int(min(255, (1.0 + 0.66) * med))
    return cv2.Canny(g8, lo, hi) > 0

def _void_mask_v2(gray, ksize=31, C=-0.02, min_size=256):
    """
    Build a void mask by treating locally *brighter* regions as 'voids' (holes in mass),
    then clean with morphology to suppress texture speckle.
    """
    g = (gray - gray.min()) / (gray.max() - gray.min() + 1e-8)
    g_blur = cv2.GaussianBlur(g, (0, 0), 3.0)

    th = cv2.adaptiveThreshold(
        (g_blur * 255).astype(np.uint8),
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        ksize | 1,
        int(C * 255),
    )
    voids = th.astype(bool)
    voids = binary_opening(voids, disk(2))
    voids = binary_closing(voids, disk(3))
    voids = remove_small_objects(voids, min_size=min_size)
    voids = remove_small_holes(voids, area_threshold=min_size)
    return voids

def void_metrics(gray):
    """
    Returns:
      V         : number of detected voids (bright regions)
      chi       : bounded mass-fragmentation proxy (negative, capped)
      AR        : aspect ratio of the largest void's bbox
      cut_depth : max interior distance of largest void (pixels)
      edges     : Canny edges (for visualization only)
      voids     : boolean mask of voids
    """
    voids = _void_mask_v2(gray)
    edges = _edges_for_viz(gray)

    lab = label(voids)
    V   = int(lab.max())

    # Largest-void features
    if V == 0:
        AR = 1.0
        cut_depth = 0.0
    else:
        props = sorted(regionprops(lab), key=lambda r: r.area, reverse=True)
        p0    = props[0]
        (minr, minc, maxr, maxc) = p0.bbox
        H = max(1, maxr - minr)
        W = max(1, maxc - minc)
        AR = max(H, W) / float(min(H, W))
        cut_depth = float(distance_transform_edt(lab == p0.label).max())

    # Bounded fragmentation proxy for 'chi' (keeps API, avoids huge magnitudes)
    # Interpret as: more mass fragments (non-void islands) → more negative.
    mass = ~voids
    n_islands = int(label(mass).max())
    chi = float(-min(128, n_islands))  # cap to keep scores stable

    return {
        "V": V,
        "chi": chi,
        "AR": float(AR),
        "cut_depth": float(cut_depth),
        "edges": edges,
        "voids": voids,
    }
```

```python

def contour_curvature_stats(gray, level=None):
    g = gaussian_filter(gray, 1.0)
    if level is None:
        level = float(g.mean())
    contours = find_contours(g, level=level)
    if not contours:
        return {"k_var":0.0, "infl_density":0.0, "n":0}
    curvs = []
    infl = 0
    for c in contours:
        if len(c) < 9:
            continue
        x = c[:,1]; y = c[:,0]
        k = 9 if 9 < len(c) else (len(c)-1 if (len(c)-1)%2==1 else len(c)-2)
        xs = savgol_filter(x, k, 2, mode='interp')
        ys = savgol_filter(y, k, 2, mode='interp')
        dx = np.gradient(xs); dy = np.gradient(ys)
        ddx = np.gradient(dx); ddy = np.gradient(dy)
        denom = (dx*dx + dy*dy)**1.5 + 1e-8
        kappa = (dx*ddy - dy*ddx) / denom  # signed curvature
        curvs.append(kappa)
        infl += int(np.sum(np.sign(kappa[:-1]) != np.sign(kappa[1:])))
    if not curvs:
        return {"k_var":0.0, "infl_density":0.0, "n":0}
    curvs = np.concatenate(curvs)
    k_var = float(np.var(curvs))
    infl_density = float(infl / max(1, curvs.size))
    return {"k_var": k_var, "infl_density": infl_density, "n": int(curvs.size)}
```

```python

def orientation_entropy(gray, bins=36):
    gx = filters.sobel_h(gray)
    gy = filters.sobel_v(gray)
    mag = np.hypot(gx, gy) + 1e-8
    ang = (np.arctan2(gy, gx) + np.pi) % np.pi
    hist, _ = np.histogram(ang, bins=bins, range=(0,np.pi), weights=mag)
    p = hist / (hist.sum() + 1e-8)
    H = -np.sum(p * np.log2(p + 1e-12))

    # More aggressive scaling to ensure output stays in 0-10 range
    H_max = np.log2(bins)
    H_normalized = (H / H_max) * 12.0  # Changed from 10.0 to 8.0
    return float(H_normalized)

def G4_occlusion_entropy_proxy(gray):
    # Proxy: higher orientation entropy ~ more competing structure (potential near/far ambiguity)
    H = orientation_entropy(gray, bins=36)
    return {"H": H}
```

```python
# --- T-junction occlusion DAG (compact reference) ---

import cv2, numpy as np, networkx as nx
from skimage import feature, segmentation, morphology
from skimage.filters import sobel

def tjunctions(gray):
    g = (gray*255).astype(np.uint8)
    e = cv2.Canny(g, 50, 150)  # tweakable
    # Harris corners to preselect interest points
    h = cv2.cornerHarris(g, 2, 3, 0.04)
    ys, xs = np.where(h > 0.01*h.max())

    # local orientation via structure tensor
    gx, gy = np.gradient(gray)
    ang = (np.arctan2(gy, gx) + np.pi) % np.pi   # [0,pi)

    T = []
    for y, x in zip(ys, xs):
        y0, y1 = max(0,y-3), min(gray.shape[0], y+4)
        x0, x1 = max(0,x-3), min(gray.shape[1], x+4)
        win = e[y0:y1, x0:x1]
        if win.sum() < 12:  # not enough strokes
            continue
        # crude: find two dominant angle modes
        aw = ang[y0:y1, x0:x1][win>0]
        if aw.size < 10:
            continue
        hist, bins = np.histogram(aw, bins=18, range=(0,np.pi))
        modes = np.argsort(hist)[-2:]
        a0 = 0.5*(bins[modes[0]] + bins[modes[0]+1])
        a1 = 0.5*(bins[modes[1]] + bins[modes[1]+1])
        # large angle difference ≈ T (vs X/Y)
        if abs(np.sin(a0 - a1)) < 0.6:  # reject near-parallel
            continue
        # Heuristic occluder: orientation with the shorter visible segment → "bar"
        # We approximate by counting pixels aligned with each mode.
        n0 = hist[modes[0]]; n1 = hist[modes[1]]
        bar = 0 if n0 < n1 else 1
        T.append((y, x, a0, a1, bar))
    return T, e

def regions_from_edges(gray, edges):
    # watershed on gradient -> labeled regions
    elev = sobel(gray)
    m = morphology.remove_small_holes(~edges.astype(bool), 64)
    lab = segmentation.watershed(elev, markers=None, mask=m)
    return lab

def occlusion_graph(gray):
    T, e = tjunctions(gray)
    lab = regions_from_edges(gray, e)
    G = nx.DiGraph()
    G.add_nodes_from(np.unique(lab)[1:])  # skip 0
    h, w = gray.shape
    for (y, x, a0, a1, bar) in T:
        # sample normals to pick near regions
        y0, x0 = int(np.clip(y,0,h-1)), int(np.clip(x,0,w-1))
        nbh = lab[max(0,y0-1):min(h,y0+2), max(0,x0-1):min(w,x0+2)]
        regs = [r for r in np.unique(nbh) if r != 0]
        if len(regs) < 2:
            continue
        # very rough: bar region occludes the others at the junction
        occluder = regs[bar % len(regs)]
        for r in regs:
            if r != occluder:
                G.add_edge(occluder, r)
    return G

def occlusion_entropy_dag(G: nx.DiGraph) -> float:
    if not G.number_of_nodes():
        return 0.0
    indeg = np.array([d for _, d in G.in_degree()], float)
    outdeg = np.array([d for _, d in G.out_degree()], float)

    def ent(v):
        p = v / (v.sum() + 1e-8)
        return float(-np.sum(p * np.log2(p + 1e-12))) if (p > 0).any() else 0.0

    H_in, H_out = ent(indeg + 1e-8), ent(outdeg + 1e-8)
    H = H_in + H_out
    # normalize to a friendly band (~0..6). You can calibrate with your corpus.
    return float(H)

def G4_occlusion_entropy_true(gray):
    G = occlusion_graph(gray)
    H = occlusion_entropy_dag(G)
    return {"H": H}

# ---- G4 switch (proxy vs true DAG) ----
USE_TRUE_G4 = False   # set False to fall back to proxy quickly

def G4_extract(gray):
    if USE_TRUE_G4:
        return G4_occlusion_entropy_true(gray)
    else:
        return G4_occlusion_entropy_proxy(gray)
```

```python
# Δ_rec helpers ---------------------------------------------------------------
import re, numpy as np
from collections import defaultdict

def _stem_for(path: str) -> str:
    n = path.split("/")[-1]
    n = re.sub(r"\s*\(\d+\)(\.\w+)?$", lambda m: m.group(1) or "", n)
    n = re.sub(r"_iter\d+(\.\w+)$", r"\1", n, flags=re.I)
    n = re.sub(r"\s+iter\s*\d+", "", n, flags=re.I)
    return n

def group_passes(paths):
    g = defaultdict(list)
    for p in paths: g[_stem_for(p)].append(p)
    for k in g: g[k] = sorted(g[k])
    return g

def _features_to_z(feats, *, z_clip=None):
    # use the exact normalizer name you have in the notebook:
    f = _normalize_feats(feats)
    z1 = _z_or_robust("G1_L", f["G1_L"]) + 0.5*_z_or_robust("G1_K", f["G1_K"])
    z2 = (_z_or_robust("G2_V", f["G2_V"]) +
          0.75*_z_or_robust("G2_cut", f["G2_cut"]) +
          0.5*_z_or_robust("G2_AR", f["G2_AR"]) +
          0.5*_z_or_robust("G2_chi", f["G2_chi"]))
    z3 = _z_or_robust("G3_kvar", f["G3_kvar"]) + _z_or_robust("G3_infl", f["G3_infl"])
    z4 = _z_or_robust("G4_H", f["G4_H"])
    z = np.array([z1, z2, z3, z4], float)
    if z_clip is not None: z = np.clip(z, -z_clip, z_clip)
    return z

def delta_rec_from_passes(paths, *, z_clip=3.5):
    zs = []
    for p in paths:
        _, feats, _ = compute_vclig_strict(p)  # strict: no REFS mutation
        zs.append(_features_to_z(feats, z_clip=z_clip))
    if len(zs) < 2: return 0.0
    zs = np.stack(zs, 0)
    diffs = np.linalg.norm(np.diff(zs, axis=0), axis=1)
    d = float(np.median(diffs))
    return max(0.0, min(4.0, d))
```

## G5 — Structural Coherence Index (SCI)

**What it measures**: Whether complexity is organized or entropic.

**Four sub-signals**:
1. **Regional Consistency** (30% weight) — Do different areas of the image have similar structural properties?
2. **Edge Alignment** (30% weight) — Do edges follow dominant angles, or scatter randomly?
3. **Scale Consistency** (25% weight) — Does detail density stay proportional across zoom levels?
4. **Rhythm/Repetition** (15% weight) — Are there systematic patterns via autocorrelation?

**Returns**: SCI composite (0–5) plus four sub-metrics.

**High SCI** → Edges align, regions are consistent, detail is proportional, rhythm is present.
**Low SCI** → High local variance, scattered orientations, detail spikes unpredictably.

**Not measured**: Whether the organization is "good composition." SCI detects systematic structure, not aesthetic merit. A perfectly organized gradient scores high; a masterful Pollock scores low. Both are valid.

```python
# --- Structural Coherence Index (SCI) ---
import numpy as np
import cv2
from scipy.signal import correlate2d
from skimage.filters import sobel

def structural_coherence_index(gray):
    """
    Measures whether complexity is organized vs. entropic.
    Returns dict with sub-metrics and composite SCI score (0-5).

    Low SCI = chaotic, entropic, scattered
    High SCI = organized, coherent, systematic
    """
    h, w = gray.shape

    # 1. Regional Consistency - divide into 4 quadrants, compare variance
    qh, qw = h // 2, w // 2
    quadrants = [
        gray[0:qh, 0:qw],
        gray[0:qh, qw:w],
        gray[qh:h, 0:qw],
        gray[qh:h, qw:w]
    ]

    # Edge density per quadrant
    q_edge_density = []
    for q in quadrants:
        edges = sobel(q)
        q_edge_density.append(float(edges.mean()))

    # Low variance across quadrants = consistent structure
    regional_consistency = 1.0 / (1.0 + np.var(q_edge_density))

    # 2. Edge Alignment - measure angular entropy of edges
    gx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    gy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    mag = np.hypot(gx, gy)
    angles = np.arctan2(gy, gx)

    # Weight by magnitude and compute histogram
    angle_bins = 36
    hist, _ = np.histogram(angles[mag > np.percentile(mag, 25)],
                          bins=angle_bins, range=(-np.pi, np.pi))

    # Entropy of angle distribution (low = aligned, high = scattered)
    p = hist / (hist.sum() + 1e-8)
    angle_entropy = -np.sum(p * np.log2(p + 1e-12))
    max_entropy = np.log2(angle_bins)
    angle_alignment = 1.0 - (angle_entropy / max_entropy)  # Invert so high = aligned

    # 3. Scale Consistency - detail density across pyramid levels
    pyramid_levels = []
    current = gray.copy()
    for _ in range(4):
        edges = sobel(current)
        pyramid_levels.append(float(edges.mean()))
        if min(current.shape) < 32:
            break
        current = cv2.pyrDown(current)

    if len(pyramid_levels) > 1:
        scale_consistency = 1.0 / (1.0 + np.std(pyramid_levels) / (np.mean(pyramid_levels) + 1e-8))
    else:
        scale_consistency = 1.0

    # 4. Rhythm/Repetition - autocorrelation strength
    # Downsample for speed
    small = cv2.resize(gray, (min(128, w), min(128, h)))
    small = (small - small.mean()) / (small.std() + 1e-8)

    # Compute autocorrelation
    autocorr = correlate2d(small, small, mode='same')
    autocorr = autocorr / autocorr.max()

    # Find peaks excluding center
    center_y, center_x = np.array(autocorr.shape) // 2
    mask = np.ones_like(autocorr, dtype=bool)
    mask[center_y-5:center_y+5, center_x-5:center_x+5] = False

    peaks = autocorr[mask] > 0.3  # Threshold for significant peaks
    rhythm_strength = float(np.sum(peaks)) / 100.0  # Normalize
    rhythm_strength = min(1.0, rhythm_strength)

    # Composite SCI score
    # Weights: regional consistency (0.3), angle alignment (0.3),
    #          scale consistency (0.25), rhythm (0.15)
    sci_raw = (0.30 * regional_consistency +
               0.30 * angle_alignment +
               0.25 * scale_consistency +
               0.15 * rhythm_strength)

    # Map to 0-5 scale (already in 0-1 range, so just scale)
    sci_score = float(np.clip(sci_raw * 5.0, 0, 5))

    return {
        "SCI": sci_score,
        "regional_consistency": float(regional_consistency),
        "angle_alignment": float(angle_alignment),
        "scale_consistency": float(scale_consistency),
        "rhythm_strength": float(rhythm_strength),
    }
```

```python
# --- Running stats & REFS (required by strict scorer) ---
class RunningStats:
    def __init__(self):
        self.values = []
    def update(self, v):
        self.values.append(float(v))
    def z(self, v):
        import numpy as np
        vals = np.array(self.values) if self.values else np.array([v])
        mu = float(vals.mean())
        sd = float(vals.std() + 1e-8)
        return (float(v) - mu) / sd

# Global refs (empty is OK; you can seed priors later)
REFS = {
    "G1_L":   RunningStats(),
    "G1_K":   RunningStats(),
    "G2_V":   RunningStats(),
    "G2_chi": RunningStats(),
    "G2_AR":  RunningStats(),
    "G2_cut": RunningStats(),
    "G3_kvar":RunningStats(),
    "G3_infl":RunningStats(),
    "G4_H":   RunningStats(),
}
```

```python
# --- Cold-start fallback & robust normalization patch ---

import numpy as np

# 1) Replace Euler number with bounded "hole ratio"
def _bounded_holes(voids):
    # holes = number of 0-islands inside the largest connected 1-region
    # simple bound: min(128, count); normalize by area later if desired
    from skimage.measure import label
    lab = label(voids)
    return min(128, int(lab.max()))

# override in void_metrics post-process
# --- VCLI-G: side-effect–free extraction + clean scoring (drop-in cell) ---
# Requires the earlier helpers already in your notebook:
# load_image, to_gray, G1_centroid_wander, void_metrics,
# contour_curvature_stats, G4_occlusion_entropy_proxy, and the REFS dict.

import numpy as np

# normalize heavy-tailed features
def _normalize_feats(feats: dict) -> dict:
    f = dict(feats)
    if "G2_chi" in f:
        f["G2_chi"] = -min(128.0, abs(float(f["G2_chi"])))  # bounded, monotone
    f["G2_cut"]  = np.log1p(max(0.0, f["G2_cut"]))
    f["G3_kvar"] = np.log1p(max(0.0, f["G3_kvar"]))
    return f

# 1) Pure feature extraction — NO updates to REFS (no side effects)
def extract_features_noupdate(image_path, sigmas=(2,4,8,16)):
    """Pure features. NO REFS mutation."""
    img  = load_image(image_path)
    gray = to_gray(img)

    g1 = G1_centroid_wander(gray, sigmas=sigmas)
    g2 = void_metrics(gray)
    g3 = contour_curvature_stats(gray)
    g4 = G4_extract(gray)
    sci_data = structural_coherence_index(gray)  # ADD THIS LINE

    return {
        "G1_L": float(g1["path_len"]),
        "G1_K": float(g1["path_curv"]),
        "G2_V": int(g2["V"]),
        "G2_chi": float(g2["chi"]),
        "G2_AR": float(g2["AR"]),
        "G2_cut": float(g2["cut_depth"]),
        "G3_kvar": float(g3["k_var"]),
        "G3_infl": float(g3["infl_density"]),
        "G4_H": float(g4["H"]),
        # ADD THESE LINES:
        "SCI": float(sci_data["SCI"]),
        "SCI_regional": float(sci_data["regional_consistency"]),
        "SCI_angle": float(sci_data["angle_alignment"]),
        "SCI_scale": float(sci_data["scale_consistency"]),
        "SCI_rhythm": float(sci_data["rhythm_strength"]),
    }

# 2) Robust z-score for cold start (falls back when refs are empty or σ≈0)
def _robust_z(x, series):
    arr = np.array(series) if len(series) else np.array([x])
    med = float(np.median(arr))
    mad = float(np.median(np.abs(arr - med))) + 1e-6
    return (float(x) - med) / (1.4826 * mad)

def _z_or_robust(key, val):
    ref = REFS[key].values
    return _robust_z(val, ref) if (len(ref) < 3 or np.std(ref) < 1e-6) else REFS[key].z(val)

# 3) Cold-start–safe scorer (0–5 band)
def vclig_score_coldstart_safe(features, *, weights=None, squash=2.5, z_clip=None, delta_rec=0.0):
    """Return (score[0..5], {'z1','z2','z3','z4','raw'}) without mutating REFS."""
    if weights is None:
        weights = dict(w1=0.25, w2=0.20, w3=0.25, w4=0.20, w5=0.10)

    f = _normalize_feats(features)

    z1 = _z_or_robust("G1_L", f["G1_L"]) + 0.5 * _z_or_robust("G1_K", f["G1_K"])
    z2_raw = (_z_or_robust("G2_V",  f["G2_V"]) +
          0.75 * _z_or_robust("G2_cut", f["G2_cut"]) +
          0.5  * _z_or_robust("G2_AR",  f["G2_AR"]) +
          0.5  * _z_or_robust("G2_chi", f["G2_chi"]))
    z2 = z2_raw / 2.5  # Normalize by sum of weights to keep scale consistent
    z3 = (_z_or_robust("G3_kvar", f["G3_kvar"]) +
          _z_or_robust("G3_infl", f["G3_infl"]))
    z4 = _z_or_robust("G4_H", f["G4_H"])

    if z_clip is not None:
        clip = lambda v: max(-z_clip, min(z_clip, v))
        z1, z2, z3, z4 = map(clip, (z1, z2, z3, z4))

    raw = (weights["w1"]*z1 + weights["w2"]*z2 +
           weights["w3"]*z3 + weights["w4"]*z4 + weights["w5"]*delta_rec)

    score = 2.5 + np.tanh(raw / float(squash)) * 2.5
    return float(np.clip(score, 0, 5)), {"z1": float(z1), "z2": float(z2), "z3": float(z3), "z4": float(z4), "raw": float(raw)}

# 4) Strict API — no REFS mutation anywhere during extraction
def compute_vclig_strict(image_path, *, sigmas=(2,4,8,16), weights=None, squash=2.5, z_clip=None, delta_rec=0.0):
    feats = extract_features_noupdate(image_path, sigmas=sigmas)
    score, zinfo = vclig_score_coldstart_safe(feats, weights=weights, squash=squash, z_clip=z_clip, delta_rec=delta_rec)
    return score, feats, zinfo
```

```python
# --- Sequential delta tracking based on upload order (NO filename parsing) ---
import pandas as pd
import numpy as np

ALL_COLS = [
    "path", "sequence_idx", "profile", "VCLI_G", "delta_step", "delta_cumulative",
    "G1_L", "G1_K",
    "G2_V", "G2_chi", "G2_AR", "G2_cut",
    "G3_kvar", "G3_infl",
    "G4_H",
    "SCI", "SCI_regional", "SCI_angle", "SCI_scale", "SCI_rhythm",
    "z_z1", "z_z2", "z_z3", "z_z4", "z_raw",
    "z1_drift", "z2_drift", "z3_drift", "z4_drift",
]

def batch_vclig_sequential(paths, profile=None):
    """
    Process images in the exact order provided.
    Calculate deltas between consecutive images (1→2, 2→3, 3→4, etc.)
    NO filename parsing, NO grouping - just sequential processing.
    """
    name = profile or VCLIG_ACTIVE_PROFILE
    cfg = VCLIG_PROFILES[name]
    zclip = cfg.get("z_clip", None)

    # Extract features and z-vectors for all images
    zs = []
    feats_list = []
    for p in paths:
        _, feats, _ = compute_vclig_strict(p)
        feats_list.append(feats)
        zs.append(_features_to_z(feats, z_clip=zclip))

    zs = np.array(zs)  # shape: (n_images, 4)

    # Calculate step-wise deltas (consecutive images)
    delta_steps = [0.0]  # First image has no delta
    if len(zs) > 1:
        diffs = np.linalg.norm(np.diff(zs, axis=0), axis=1)  # L2 distance
        delta_steps.extend(diffs.tolist())

    # Calculate per-component drift
    z_drifts = []
    if len(zs) > 1:
        component_diffs = np.diff(zs, axis=0)  # shape: (n-1, 4)
        z_drifts = component_diffs.tolist()

    # Build rows
    rows = []
    cumulative_delta = 0.0

    for i, (p, feats) in enumerate(zip(paths, feats_list)):
        try:
            delta_step = delta_steps[i]
            cumulative_delta += delta_step

            # Score WITHOUT delta (delta is just metadata, doesn't affect score)
            s, _, z = compute_vclig_strict(
                p,
                weights=cfg["weights"],
                squash=cfg["squash"],
                z_clip=zclip,
                delta_rec=0.0  # Keep at 0 - deltas are observational only
            )

            s = float(np.clip(s + cfg.get("bias", 0.0), 0, 5))

            row = {
                "path": p,
                "sequence_idx": i + 1,  # 1-indexed for readability
                "profile": name,
                "VCLI_G": round(float(s), 3),
                "delta_step": round(float(delta_step), 4),
                "delta_cumulative": round(float(cumulative_delta), 4),
                **{k: float(v) for k, v in feats.items()},
                "z_z1": float(z["z1"]),
                "z_z2": float(z["z2"]),
                "z_z3": float(z["z3"]),
                "z_z4": float(z["z4"]),
                "z_raw": float(z["raw"]),
            }

            # Add per-component drift (if not first image)
            if i > 0 and z_drifts:
                drift = z_drifts[i - 1]
                row.update({
                    "z1_drift": round(float(drift[0]), 4),
                    "z2_drift": round(float(drift[1]), 4),
                    "z3_drift": round(float(drift[2]), 4),
                    "z4_drift": round(float(drift[3]), 4),
                })
            else:
                row.update({
                    "z1_drift": np.nan,
                    "z2_drift": np.nan,
                    "z3_drift": np.nan,
                    "z4_drift": np.nan,
                })

            rows.append(row)

        except Exception as e:
            rows.append({
                "path": p,
                "sequence_idx": i + 1,
                "profile": name,
                "error": str(e)
            })

    return rows

def results_table_sequential(paths, *, profile=None, csv_path=None):
    """
    Generate results with sequential delta tracking.
    Images are processed in the order provided - no filename parsing.
    """
    data = batch_vclig_sequential(paths, profile=profile)
    df = pd.DataFrame(data)

    # Ensure all columns exist and order them
    for c in ALL_COLS:
        if c not in df.columns:
            df[c] = np.nan

    df = df[ALL_COLS + [c for c in df.columns if c not in ALL_COLS]]

    # Round numerics
    for c in df.columns:
        if pd.api.types.is_numeric_dtype(df[c]):
            df[c] = df[c].astype(float).round(4)

    display(df)

    if csv_path:
        df.to_csv(csv_path, index=False)
        print(f"Saved to {csv_path}")

    return df
```

```python
# --- Sequential delta tracking based on upload order (NO filename parsing) ---
import pandas as pd
import numpy as np

ALL_COLS = [
    "sequence_idx", "path", "profile", "VCLI_G", "delta_step", "delta_cumulative",
    "G1_L", "G1_K",
    "G2_V", "G2_chi", "G2_AR", "G2_cut",
    "G3_kvar", "G3_infl",
    "G4_H",
    "SCI", "SCI_regional", "SCI_angle", "SCI_scale", "SCI_rhythm",
    "z_z1", "z_z2", "z_z3", "z_z4", "z_raw",
    "z1_drift", "z2_drift", "z3_drift", "z4_drift",
]

def batch_vclig_sequential(paths, profile=None):
    """
    Process images in the exact order provided.
    Calculate deltas between consecutive images (1→2, 2→3, 3→4, etc.)
    NO filename parsing, NO grouping - just sequential processing.
    """
    name = profile or VCLIG_ACTIVE_PROFILE
    cfg = VCLIG_PROFILES[name]
    zclip = cfg.get("z_clip", None)

    # Extract features and z-vectors for all images
    zs = []
    feats_list = []
    for p in paths:
        _, feats, _ = compute_vclig_strict(p)
        feats_list.append(feats)
        zs.append(_features_to_z(feats, z_clip=zclip))

    zs = np.array(zs)  # shape: (n_images, 4)

    # Calculate step-wise deltas (consecutive images)
    delta_steps = [0.0]  # First image has no delta
    if len(zs) > 1:
        diffs = np.linalg.norm(np.diff(zs, axis=0), axis=1)  # L2 distance
        delta_steps.extend(diffs.tolist())

    # Calculate per-component drift
    z_drifts = []
    if len(zs) > 1:
        component_diffs = np.diff(zs, axis=0)  # shape: (n-1, 4)
        z_drifts = component_diffs.tolist()

    # Build rows
    rows = []
    cumulative_delta = 0.0

    for i, (p, feats) in enumerate(zip(paths, feats_list)):
        try:
            delta_step = delta_steps[i]
            cumulative_delta += delta_step

            # Score WITHOUT delta (delta is just metadata, doesn't affect score)
            s, _, z = compute_vclig_strict(
                p,
                weights=cfg["weights"],
                squash=cfg["squash"],
                z_clip=zclip,
                delta_rec=0.0  # Keep at 0 - deltas are observational only
            )

            s = float(np.clip(s + cfg.get("bias", 0.0), 0, 5))

            row = {
                "path": p,
                "sequence_idx": i + 1,  # 1-indexed for readability
                "profile": name,
                "VCLI_G": round(float(s), 3),
                "delta_step": round(float(delta_step), 4),
                "delta_cumulative": round(float(cumulative_delta), 4),
                **{k: float(v) for k, v in feats.items()},
                "z_z1": float(z["z1"]),
                "z_z2": float(z["z2"]),
                "z_z3": float(z["z3"]),
                "z_z4": float(z["z4"]),
                "z_raw": float(z["raw"]),
            }

            # Add per-component drift (if not first image)
            if i > 0 and z_drifts:
                drift = z_drifts[i - 1]
                row.update({
                    "z1_drift": round(float(drift[0]), 4),
                    "z2_drift": round(float(drift[1]), 4),
                    "z3_drift": round(float(drift[2]), 4),
                    "z4_drift": round(float(drift[3]), 4),
                })
            else:
                row.update({
                    "z1_drift": np.nan,
                    "z2_drift": np.nan,
                    "z3_drift": np.nan,
                    "z4_drift": np.nan,
                })

            rows.append(row)

        except Exception as e:
            rows.append({
                "path": p,
                "sequence_idx": i + 1,
                "profile": name,
                "error": str(e)
            })

    return rows

def results_table_sequential(paths, *, profile=None, csv_path=None):
    """
    Generate results with sequential delta tracking.
    Images are processed in the order provided - no filename parsing.
    """
    data = batch_vclig_sequential(paths, profile=profile)
    df = pd.DataFrame(data)

    # Ensure all columns exist and order them
    for c in ALL_COLS:
        if c not in df.columns:
            df[c] = np.nan

    df = df[ALL_COLS + [c for c in df.columns if c not in ALL_COLS]]

    # Round numerics
    for c in df.columns:
        if pd.api.types.is_numeric_dtype(df[c]):
            df[c] = df[c].astype(float).round(4)

    display(df)

    if csv_path:
        df.to_csv(csv_path, index=False)
        print(f"Saved to {csv_path}")

    return df
```

```python
# @title 3. Signal Fingerprints (Radar Chart)
import matplotlib.pyplot as plt
import numpy as np
from math import pi

# SELECT IMAGES TO COMPARE (change indices or paths as needed)
# Option 1: By index
compare_indices = [0, 1, 2, 3]  # First three images
# Option 2: By finding highest/lowest
# compare_indices = [df['VCLI_G'].idxmax(), df['VCLI_G'].idxmin()]

image_paths = df.iloc[compare_indices]['path'].tolist()

# Setup radar chart
categories = ['z1\n(Wander)', 'z2\n(Void)', 'z3\n(Torque)', 'z4\n(Occlusion)']
N = len(categories)
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(9, 9), subplot_kw=dict(projection='polar'))

colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#95E1D3']

for idx, path in enumerate(image_paths):
    row = df[df['path'] == path].iloc[0]
    values = [row['z_z1'], row['z_z2'], row['z_z3'], row['z_z4']]

    # Normalize to 0-1 range for visual clarity
    max_z = max(abs(min(values)), abs(max(values))) if max(abs(min(values)), abs(max(values))) > 0 else 1
    values_norm = [(v / max_z + 1) / 2 for v in values]
    values_norm += values_norm[:1]

    img_label = path.split('/')[-1][:20] + '...' if len(path.split('/')[-1]) > 20 else path.split('/')[-1]
    ax.plot(angles, values_norm, 'o-', linewidth=2.5,
            label=f"{img_label} ({row['VCLI_G']:.2f})", color=colors[idx % len(colors)])
    ax.fill(angles, values_norm, alpha=0.15, color=colors[idx % len(colors)])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, size=11)
ax.set_ylim(0, 1)
ax.set_yticks([0.25, 0.5, 0.75, 1.0])
ax.set_yticklabels(['', '0.5', '', '1.0'], size=9)
ax.set_title('Signal Fingerprints (Normalized)', size=14, weight='bold', pad=25)
ax.legend(loc='upper right', bbox_to_anchor=(1.35, 1.1), fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Reading the 2D Space (VCLI-G × SCI)

When analyzing images, consider their coordinates:

**Quadrant interpretations**:

| SCI ↓ / VCLI-G → | **Low (0-2.5)** | **High (2.5-5.0)** |
|------------------|-----------------|-------------------|
| **High (2.5-5)** | Resolved, clear, organized | Earned tension, deliberate complexity |
| **Low (0-2.5)**  | Default simple, unstructured | Chaotic, emergent, experimental |

**Movement across the space**:
- An artist refining a sketch might move from *low SCI → high SCI* (organizing chaos)
- A model "playing it safe" clusters in *low VCLI-G, high SCI*
- Overfitting to texture noise pushes toward *high VCLI-G, low SCI*
- Master works often live in *high VCLI-G, high SCI* (but not always!)

**The space is descriptive, not prescriptive.** Where an image should live depends on context, medium, and intent.

```python
# ----- VCLI-G × SCI "waterfall" view (drop-in) -----
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def _mahalanobis_radius(X, mu=None, Sinv=None):
    if mu is None: mu = X.mean(axis=0)
    if Sinv is None:
        S = np.cov(X, rowvar=False)
        Sinv = np.linalg.pinv(S)
    d = X - mu
    r = np.sqrt(np.einsum('ij,jk,ik->i', d, Sinv, d))
    return r, mu, Sinv

def _warp_radius(r, method="power", **kw):
    if method == "power":
        gamma = float(kw.get("gamma", 1.6))
        return np.power(r, gamma)
    elif method == "naka":
        n = float(kw.get("n", 3.0))
        sigma = float(kw.get("sigma", np.median(r)))
        Rmax = float(kw.get("Rmax", r.max()))
        r_n = np.power(r, n)
        return Rmax * (r_n / (r_n + np.power(sigma, n) + 1e-12))
    else:
        raise ValueError("method must be 'power' or 'naka'")

def warp_scatter(df, method="naka", anisotropic=False, **kw):
    """
    df: DataFrame with columns 'SCI' and 'VCLI_G'
    method: 'power' or 'naka'  (naka = contrast-like sigmoid)
    anisotropic: if True, apply different warps per axis
    kw: parameters (power: gamma=1.6; naka: n=3.0, sigma='median', Rmax=None)
    Returns a copy of df with columns: 'SCI_warp','VCLI_warp'
    """
    out = df.copy()
    X = out[["SCI","VCLI_G"]].to_numpy().astype(float)
    r, mu, Sinv = _mahalanobis_radius(X)

    if not anisotropic:
        rprime = _warp_radius(r, method, **kw)
        scale = np.divide(rprime, r, out=np.zeros_like(rprime), where=r>1e-12)
        Xw = mu + (X - mu) * scale[:, None]
        out["SCI_warp"], out["VCLI_warp"] = Xw[:,0], Xw[:,1]
    else:
        # optional: different “falls” per axis (e.g., stronger on VCLI_G)
        sci = X[:,0]; vcl = X[:,1]
        r_sci = np.abs(sci - sci.mean()) / (sci.std() + 1e-8)
        r_vcl = np.abs(vcl - vcl.mean()) / (vcl.std() + 1e-8)
        out["SCI_warp"]   = sci.mean() + np.sign(sci - sci.mean()) * _warp_radius(r_sci, method, **kw) * (sci.std() + 1e-8)
        out["VCLI_warp"]  = vcl.mean() + np.sign(vcl - vcl.mean()) * _warp_radius(r_vcl, method, **kw) * (vcl.std() + 1e-8)

    return out

def plot_raw_vs_warp(df_raw, df_warp, title="", cmap=None):
    fig, axes = plt.subplots(1, 2, figsize=(10,4), dpi=140)
    for ax, data, ttl in [(axes[0], df_raw, "Raw"), (axes[1], df_warp, title)]:
        X = data[["SCI","VCLI_G"]].to_numpy() if "SCI_warp" not in data else data[["SCI_warp","VCLI_warp"]].to_numpy()
        ax.scatter(X[:,0], X[:,1], s=22)
        ax.set_xlim(0,5); ax.set_ylim(0,5)
        ax.axvline(2.5, ls="--", lw=1); ax.axhline(2.5, ls="--", lw=1)
        ax.set_xlabel("SCI (Structural Coherence Index)")
        ax.set_ylabel("VCLI-G (Visual Cognitive Load)")
        ax.set_title(ttl)
        ax.text(0.2, 4.7, "Chaotic\nComplexity", fontsize=8, alpha=0.6)
        ax.text(4.8, 4.7, "Earned\nTension", fontsize=8, alpha=0.6, ha="right")
        ax.text(0.2, 0.3, "Default\nSimple", fontsize=8, alpha=0.6)
        ax.text(4.8, 0.3, "Resolved\nClarity", fontsize=8, alpha=0.6, ha="right")
    plt.tight_layout(); plt.show()

# --- Example usage (assumes you already have a results DataFrame 'df' with SCI/VCLI_G) ---
# df_warp = warp_scatter(df, method="naka", n=4.0, sigma=np.median(r)/3)
# plot_raw_vs_warp(df, df_warp, title="Naka–Rushton (n=4.0, σ/3)")
# df_warp = warp_scatter(df, method="power", gamma=1.6)     # power-law waterfall
# plot_raw_vs_warp(df, df_warp, title="Power-law (γ=1.6)")
```

```python
# ===== B-Table: raw vs warped numbers (ready-to-run cell) =====
import numpy as np
import pandas as pd

# --- minimal helpers (safe to re-run; no-op if you already defined them) ---
def _mahalanobis_radius(X, mu=None, Sinv=None):
    if mu is None: mu = X.mean(axis=0)
    if Sinv is None:
        S = np.cov(X, rowvar=False)
        Sinv = np.linalg.pinv(S)
    d = X - mu
    r = np.sqrt(np.einsum('ij,jk,ik->i', d, Sinv, d))
    return r, mu, Sinv

def _warp_radius(r, method="naka", **kw):
    if method == "power":
        gamma = float(kw.get("gamma", 1.8))
        return np.power(r, gamma)
    elif method == "naka":
        n     = float(kw.get("n", 4.0))
        sigma = kw.get("sigma", None)
        if sigma is None: sigma = float(np.median(r))
        Rmax  = float(kw.get("Rmax", r.max() if r.size else 1.0))
        r_n = np.power(r, n)
        s_n = np.power(sigma, n)
        return Rmax * (r_n / (r_n + s_n + 1e-12))
    else:
        raise ValueError("method must be 'naka' or 'power'")

def _warp_points(X, mu, r, rprime):
    scale = np.divide(rprime, r, out=np.zeros_like(rprime), where=r>1e-12)
    return mu + (X - mu) * scale[:, None]

def build_b_table(
    df,
    method="naka",
    n=4.0,
    sigma_policy="P40",
    Rmax_policy="P95",
    gamma=1.8,
    round_to=3,
    save_csv_path=None
):
    """
    Returns a DataFrame with raw vs warped coordinates and deltas.
    method='naka' (n, sigma_policy, Rmax_policy) or 'power' (gamma).
    sigma_policy/Rmax_policy: 'median', 'PXX', or numeric.
    """

    # 1) extract raw points
    df_in = df.copy()
    X = df_in[["SCI","VCLI_G"]].to_numpy().astype(float)

    # 2) cohort stats and radius
    r, mu, Sinv = _mahalanobis_radius(X)

    # 3) choose warp params
    if method == "naka":
        # interpret policies
        if isinstance(sigma_policy, str) and sigma_policy.upper().startswith("P"):
            p = float(sigma_policy[1:])
            sigma = np.percentile(r, p)
        elif isinstance(sigma_policy, str) and sigma_policy.lower() == "median":
            sigma = float(np.median(r))
        else:
            sigma = float(sigma_policy)

        if isinstance(Rmax_policy, str) and Rmax_policy.upper().startswith("P"):
            p = float(Rmax_policy[1:])
            Rmax = np.percentile(r, p)
        else:
            Rmax = float(Rmax_policy)

        rprime = _warp_radius(r, method="naka", n=n, sigma=sigma, Rmax=Rmax)
        Xw = _warp_points(X, mu, r, rprime)
        meta = {"method":"naka","n":n,"sigma_policy":sigma_policy,"Rmax_policy":Rmax_policy,
                "sigma":float(sigma),"Rmax":float(Rmax)}
    else:
        rprime = _warp_radius(r, method="power", gamma=gamma)
        Xw = _warp_points(X, mu, r, rprime)
        meta = {"method":"power","gamma":gamma}

    # 4) assemble B-table
    out = df_in.copy()
    out["SCI_warp"]   = Xw[:,0]
    out["VCLI_warp"]  = Xw[:,1]
    out["r"]          = r
    out["r_prime"]    = rprime
    out["ΔSCI_view"]  = out["SCI_warp"]  - out["SCI"]
    out["ΔVCLI_view"] = out["VCLI_warp"] - out["VCLI_G"]
    out["Δr_view"]    = out["r_prime"]   - out["r"]

    # quadrant labels (raw/warped, for quick sanity)
    def quadrant(x,y):
        return ("Earned" if x>=2.5 and y>=2.5 else
                "Resolved" if x>=2.5 and y<2.5 else
                "Chaotic"  if x<2.5  and y>=2.5 else
                "Default")
    out["quad_raw"]  = [quadrant(s,v) for s,v in out[["SCI","VCLI_G"]].to_numpy()]
    out["quad_view"] = [quadrant(s,v) for s,v in out[["SCI_warp","VCLI_warp"]].to_numpy()]

    # nice ordering
    cols = [
        "sequence_idx","path","profile",
        "SCI","VCLI_G",
        "SCI_warp","VCLI_warp",
        "ΔSCI_view","ΔVCLI_view",
        "r","r_prime","Δr_view",
        "quad_raw","quad_view"
    ]
    for c in cols:
        if c not in out.columns:  # tolerate missing path/profile in some runs
            out[c] = np.nan
    out = out[cols]

    # rounding + optional save
    out = out.round(round_to)
    if save_csv_path:
        out.to_csv(save_csv_path, index=False)

    return out, meta

# === RUN: pick a warp and render the B-table ===
# Option 1: adaptive Naka–Rushton (recommended for physical-neutral)
btable, meta = build_b_table(
    df,
    method="naka",
    n=4.0,
    sigma_policy="P40",
    Rmax_policy="P95",
    round_to=3,
    save_csv_path=None  # e.g., "btable_physical_neutral.csv"
)
print("Warp meta:", meta)
display(btable)

# # Option 2: anisotropic power (if you want the “Option B” feel numerically)
# # (This builds the view with power γ and then reports the same columns.)
# btable, meta = build_b_table(df, method="power", gamma=1.8, round_to=3)
# print("Warp meta:", meta)
# display(btable)
```

 Residual Contrast Expansion (RCE) - Fingers

```python
# =====================  VCLI-G × SCI — Waterfall View (drop-in)  =====================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---------- helpers ----------
def _mah_r(X):
    S = np.cov(X, rowvar=False)
    Sinv = np.linalg.pinv(S)
    mu = X.mean(axis=0)
    d = X - mu
    r = np.sqrt(np.einsum('ij,jk,ik->i', d, Sinv, d))
    return r, mu, Sinv

def _anisotropic_power(X, sci_gamma=1.2, vcli_gamma=2.0):
    """z-score each axis, raise |z| to gamma (signed), then unscale."""
    sci, vcl = X[:,0], X[:,1]
    zs = (sci - sci.mean())/(sci.std()+1e-8)
    zv = (vcl - vcl.mean())/(vcl.std()+1e-8)
    zs_w = np.sign(zs) * (np.abs(zs) ** sci_gamma)
    zv_w = np.sign(zv) * (np.abs(zv) ** vcli_gamma)
    Xw = np.column_stack([
        sci.mean() + zs_w*(sci.std()+1e-8),
        vcl.mean() + zv_w*(vcl.std()+1e-8)
    ])
    return Xw

def _naka_rushton(X, n=3.5, sigma_pct=35, rmax_pct=95):
    """Adaptive radial compression/expansion in Mahalanobis space."""
    r, mu, _ = _mah_r(X)
    sigma = np.percentile(r, sigma_pct)
    Rmax  = np.percentile(r, rmax_pct)
    rn, sn = r**n, sigma**n
    rprime = Rmax * (rn/(rn+sn+1e-12))
    scale  = np.divide(rprime, r, out=np.zeros_like(rprime), where=r>1e-12)
    return mu + (X - mu)*scale[:,None]

def _powerlaw_iso(X, gamma=1.6):
    r, mu, _ = _mah_r(X)
    rprime = r**gamma
    scale  = np.divide(rprime, r, out=np.zeros_like(rprime), where=r>1e-12)
    return mu + (X - mu)*scale[:,None]

def _finger_quantile(Xw, bins=4, axis=1, blend=0.30):
    """Subtle equalization within SCI bins; keeps 70% original by default."""
    a = Xw.copy()
    x = a[:,0]; y = a[:,1]
    edges = np.linspace(x.min(), x.max(), bins+1)
    for i in range(bins):
        m = (x>=edges[i]) & (x<=edges[i+1])
        if m.sum() < 4:
            continue
        ym = y[m]
        q  = np.argsort(ym); ranks = np.empty_like(q); ranks[q] = np.arange(q.size)
        cdf = (ranks + 0.5) / q.size
        target = np.quantile(ym, cdf)
        y[m] = (1.0-blend)*ym + blend*target
    a[:,1] = y
    return a

def _finger_residual_expand(Xw, bins=5, strength=1.6, clip_sigma=3.0):
    """
    Visible fingers: expand vertical residuals within SCI bands while
    preserving each band’s mean & local slope.
    """
    a = Xw.copy()
    x, y = a[:,0], a[:,1]
    edges = np.linspace(x.min(), x.max(), bins+1)
    for i in range(bins):
        m = (x>=edges[i]) & (x<=edges[i+1])
        if m.sum() < 4:
            continue
        xb, yb = x[m], y[m]
        xm = xb.mean()
        Xb = np.column_stack([np.ones_like(xb), (xb - xm)])
        coef, *_ = np.linalg.lstsq(Xb, yb, rcond=None)
        y_pred = Xb @ coef
        resid  = yb - y_pred
        s = np.std(resid)
        if s < 1e-8:
            continue
        r = np.clip(resid/s, -clip_sigma, clip_sigma)
        y[m] = y_pred + (r*strength)*s
    a[:,1] = y
    return a

# ---------- main ----------
def waterfall_view(
    df,
    preset="physical_neutral",      # 'physical_neutral' | 'ai_conservative' | 'balanced_plus'
    finger=None,                    # None | 'quantile' | 'residual'
    finger_bins=5,
    finger_strength=1.6,            # for residual mode
    show_table=True,
    use_precomputed=False,          # if True and SCI_warp/VCLI_warp exist, use them as-is
    return_df=True,
    round_to=3
):
    """
    Plot Raw vs View and optional B-table. Returns (df_out, title).
    df must contain columns: 'SCI', 'VCLI_G'. If use_precomputed=True and
    'SCI_warp','VCLI_warp' exist, those will drive the right panel/table.
    """
    # --- base matrix ---
    X = df[["SCI","VCLI_G"]].to_numpy().astype(float)

    # --- choose view ---
    title = ""
    if use_precomputed and {"SCI_warp","VCLI_warp"} <= set(df.columns):
        Xw = df[["SCI_warp","VCLI_warp"]].to_numpy().astype(float)
        title = "Precomputed view (from table)"
    else:
        if preset == "physical_neutral":
            Xw = _anisotropic_power(X, sci_gamma=1.2, vcli_gamma=2.0)
            title = "Power (anisotropic γ: SCI=1.2, VCLI=2.0)"
        elif preset == "ai_conservative":
            Xw = _naka_rushton(X, n=3.5, sigma_pct=35, rmax_pct=95)
            title = "Naka–Rushton (n=3.5, σ=P35, Rmax=P95)"
        elif preset == "balanced_plus":
            Xw = _powerlaw_iso(X, gamma=1.6)
            title = "Power-law (γ=1.6)"
        else:
            raise ValueError("Unknown preset")

    # --- optional fingers ---
    if finger is not None:
        if finger == "quantile":
            Xw = _finger_quantile(Xw, bins=finger_bins, axis=1, blend=0.30)
            title += f" + quantile fingers (bins={finger_bins}, blend=0.30)"
        elif finger == "residual":
            Xw = _finger_residual_expand(Xw, bins=finger_bins, strength=finger_strength)
            title += f" + residual fingers (bins={finger_bins}, strength={finger_strength})"
        else:
            raise ValueError("finger must be None | 'quantile' | 'residual'")

    # --- plot ---
    plt.close('all')
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.8), dpi=140)

    def _decorate(ax, ttl):
        ax.set_xlim(0,5); ax.set_ylim(0,5)
        ax.axvline(2.5, ls="--", lw=1); ax.axhline(2.5, ls="--", lw=1)
        ax.set_xlabel("SCI (Structural Coherence Index)")
        ax.set_ylabel("VCLI-G (Visual Cognitive Load)")
        ax.set_title(ttl)
        ax.text(0.2,4.7,"Chaotic\nComplexity", fontsize=8, alpha=0.6)
        ax.text(4.8,4.7,"Earned\nTension", fontsize=8, alpha=0.6, ha="right")
        ax.text(0.2,0.3,"Default\nSimple", fontsize=8, alpha=0.6)
        ax.text(4.8,0.3,"Resolved\nClarity", fontsize=8, alpha=0.6, ha="right")

    axes[0].scatter(X[:,0],  X[:,1],  s=36)
    _decorate(axes[0], "Raw")

    axes[1].scatter(Xw[:,0], Xw[:,1], s=36)
    _decorate(axes[1], title)

    plt.tight_layout()
    plt.show()

    # --- table ---
    df_out = df.copy()
    df_out["SCI_warp"]   = Xw[:,0]
    df_out["VCLI_warp"]  = Xw[:,1]
    df_out["ΔSCI_view"]  = df_out["SCI_warp"]  - df_out["SCI"]
    df_out["ΔVCLI_view"] = df_out["VCLI_warp"] - df_out["VCLI_G"]

    if show_table:
        cols = ["sequence_idx","path","profile","SCI","VCLI_G","SCI_warp","VCLI_warp","ΔSCI_view","ΔVCLI_view"]
        display(df_out.filter(cols).sort_values("sequence_idx").round(round_to))

    return (df_out, title) if return_df else None

# ===================== examples =====================
# df_view, ttl = waterfall_view(df, preset="physical_neutral", finger=None)                       # Option B clean
# df_view, ttl = waterfall_view(df, preset="physical_neutral", finger="residual", finger_strength=1.7)  # visible fingers
# df_view, ttl = waterfall_view(df, preset="ai_conservative")                                     # Option A
# df_view, ttl = waterfall_view(df, preset="balanced_plus")                                       # Option C
# df_view, ttl = waterfall_view(df, use_precomputed=True)                                         # use SCI_warp/VCLI_warp already in df
```

```python
# --- Correlation heatmap (robust & compact) ---
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def plot_correlation_heatmap(
    df: pd.DataFrame,
    figsize=(10, 8),
    annot=True,
    annot_size=7,          # numbers inside cells
    tick_size=8,           # axis tick label size
    cbar_tick_size=8,      # colorbar tick size
    fmt=".2f",             # number format in cells
    cmap="RdBu_r"
):
    """
    Builds a correlation matrix across VCLI_G, SCI, z1..z4 and geometric submetrics
    (G1_L, G1_K, G2_V, G2_chi, G2_AR, G2_cut, G3_kvar, G3_infl, G4_H) if present.
    Handles missing/constant columns gracefully.
    """

    # Normalize possible alternate names to z1..z4 (optional)
    rename_map = {
        "z_z1":"z1","z_z2":"z2","z_z3":"z3","z_z4":"z4",
    }
    work = df.rename(columns={k:v for k,v in rename_map.items() if k in df.columns})

    # Candidate columns (include only those that actually exist)
    candidates = [
        "VCLI_G","SCI","delta_step","delta_cumulative","delta_rec",  # delta_rec optional
        "z1","z2","z3","z4",
        "G1_L","G1_K","G2_V","G2_chi","G2_AR","G2_cut","G3_kvar","G3_infl","G4_H",
    ]
    cols = [c for c in candidates if c in work.columns]

    if len(cols) < 3:
        print("Not enough overlapping columns to compute a correlation matrix.")
        print("Available:", sorted(work.columns))
        return

    # Keep numeric, drop all-NaN columns
    num = work[cols].select_dtypes(include="number").dropna(how="all", axis=1)
    if num.shape[1] < 3:
        print("Not enough numeric columns after cleaning for correlation.")
        display(num.head())
        return

    # Drop constant columns (variance == 0) to avoid NaNs in corr
    nunique = num.nunique()
    keep = nunique[nunique > 1].index
    num = num[keep]
    if num.shape[1] < 3:
        print("Not enough non-constant numeric columns for correlation.")
        display(num.head())
        return

    corr = num.corr(numeric_only=True)

    plt.figure(figsize=figsize)
    ax = sns.heatmap(
        corr, annot=annot, fmt=fmt,
        annot_kws={"size": annot_size},
        vmin=-1, vmax=1, cmap=cmap,
        square=True, linewidths=.5, cbar_kws={"shrink": 0.85}
    )
    ax.set_title("Correlation Matrix: VCLI_G × geometric signals",
                 fontsize=14, fontweight="bold", pad=20)

    # Shrink tick labels + colorbar labels
    ax.tick_params(axis="x", labelsize=tick_size, rotation=90)
    ax.tick_params(axis="y", labelsize=tick_size)
    if ax.collections and ax.collections[0].colorbar:
        ax.collections[0].colorbar.ax.tick_params(labelsize=cbar_tick_size)

    plt.tight_layout()
    plt.show()

# Call
plot_correlation_heatmap(df)
```

⚠️ Interpreting Results — Common Pitfalls

Because VCLI-G measures perceptual load rather than polish or intent, it’s possible to misread a high score if you’re not careful. A few common edge cases:

False Positives (Fake Friction):
Some AI-generated images produce high scores simply because they are messy or incoherent — not because they contain deliberate tension. The system sees “delay,” but the viewer is really just confused. Always check whether friction feels intentional (driven by form, space, or symbolism) or accidental (caused by noise or artifacts).

False Negatives (Deceptive Clarity):
Highly structured or elegant works can score lower because they resolve quickly — even if they’re artistically profound. A 2.5 score doesn’t mean the image is “lesser”; it just means the viewer doesn’t need to work hard to parse it.

Overfitting to Complexity:
More complexity ≠ higher consequence. A crowded or chaotic image may overwhelm the viewer (high load) without ever producing meaningful recursion. The best high-VCLI images sustain focused ambiguity, not just confusion.

Structural vs. Perceptual Drift:
Remember: VCLI-G is blind to compositional correctness. A work with poor spatial logic might score high, while a beautifully structured image might score low. That’s why pairing VCLI with Sketcher (structure) or Artist Lens (poise and delay) gives a more complete picture.

✅ Rule of thumb:
A high VCLI-G score means the image demands attention — but only context, critique, and comparison will tell you whether that demand is earned or incidental. Use it as one dimension of analysis, not a verdict.

```python
# === Iterative Add-On: Composite Behavior Metrics (0910G-safe, no df edits) ===
# Drop this cell AFTER your base scoring cells.
# It does NOT rename or mutate your df; it only reads from it and computes outputs.

import numpy as np, pandas as pd, json
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Optional, Dict

# ---- helpers: safe column access (tries multiple aliases, never mutates df) ----
def _col(df: pd.DataFrame, *names: str, required: bool = True, default=None):
    for n in names:
        if n in df.columns:
            return pd.to_numeric(df[n], errors="coerce").to_numpy()
    if required:
        missing = " / ".join(names)
        raise KeyError(f"Required column not found: {missing}")
    if default is None:
        return np.full(len(df), np.nan, dtype=float)
    return np.full(len(df), default, dtype=float)

def _iter_axis(df: pd.DataFrame):
    # prefer an explicit iteration column if you have it; else use 0..N-1
    if "iter" in df.columns:
        return pd.to_numeric(df["iter"], errors="coerce").to_numpy()
    if "sequence_idx" in df.columns:
        return pd.to_numeric(df["sequence_idx"], errors="coerce").to_numpy()
    return np.arange(len(df), dtype=float)

def _trapezoid_area(y: np.ndarray) -> float:
    if y.size == 0: return 0.0
    if y.size == 1: return float(y[0])
    return float(np.trapz(y, dx=1.0) / y.size)

def _linear_slope(y: np.ndarray) -> float:
    n = y.size
    if n < 2: return 0.0
    x = np.arange(n, dtype=float)
    vx = np.var(x)
    if vx == 0.0: return 0.0
    return float(np.cov(x, y, bias=True)[0, 1] / vx)

def _first_crossing(y: np.ndarray) -> Optional[int]:
    if y.size < 2: return None
    s = np.sign(y)
    idx = np.where(s[:-1] * s[1:] <= 0)[0]
    return int(idx[0] + 1) if idx.size else None

def _stability_length(y1: np.ndarray, y2: np.ndarray, eps: float = 0.05) -> int:
    if y1.size == 0 or y2.size == 0: return 0
    dy1 = np.abs(np.diff(y1, prepend=y1[:1]))
    dy2 = np.abs(np.diff(y2, prepend=y2[:1]))
    ok = (dy1 < eps) & (dy2 < eps)
    run = 0
    for flag in ok[::-1]:
        if flag: run += 1
        else: break
    return int(run)

@dataclass
class SequenceMetrics:
    H_vcli: float
    H_sci: float
    i_peak: int
    rise_slope: float
    decay_slope: float
    x_over: Optional[int]
    stab_len: int
    epsilon: float
    def to_dict(self) -> Dict:
        return {
            "H_vcli": float(self.H_vcli),
            "H_sci": float(self.H_sci),
            "i_peak": int(self.i_peak),
            "rise_slope": float(self.rise_slope),
            "decay_slope": float(self.decay_slope),
            "x_over": None if self.x_over is None else int(self.x_over),
            "stab_len": int(self.stab_len),
            "epsilon": float(self.epsilon),
        }

def compute_sequence_metrics_from_df(df: pd.DataFrame, epsilon: float = 0.05) -> SequenceMetrics:
    # Core series (required)
    v = _col(df, "VCLI_G")                     # delay index 0–5
    s = _col(df, "SCI")                        # organization 0–5

    # Derivations
    H_vcli = _trapezoid_area(v)
    H_sci  = _trapezoid_area(s)

    i_peak = int(np.argmax(v)) if v.size else 0
    rise   = v[: i_peak + 1] if i_peak >= 1 else v[:1]
    decay  = v[i_peak :] if i_peak < v.size else v[-1:]

    rise_slope  = _linear_slope(rise)
    decay_slope = _linear_slope(decay)
    x_over      = _first_crossing(v - s)
    stab_len    = _stability_length(v, s, eps=epsilon)

    return SequenceMetrics(H_vcli, H_sci, i_peak, rise_slope, decay_slope, x_over, stab_len, epsilon)

# ---- plotting (each chart its own figure; no explicit colors) ----
def plot_vcli_sci_timeline(df: pd.DataFrame, metrics: SequenceMetrics) -> None:
    it = _iter_axis(df)
    v  = _col(df, "VCLI_G")
    s  = _col(df, "SCI")

    plt.figure(figsize=(8, 4.5))
    plt.plot(it, v, label="VCLI-G")
    plt.plot(it, s, label="SCI")
    if len(it) and metrics.i_peak < len(it):
        plt.scatter([it[metrics.i_peak]], [v[metrics.i_peak]])
        plt.text(it[metrics.i_peak], v[metrics.i_peak], "peak", ha="left", va="bottom")
    if metrics.stab_len > 1:
        start = max(len(it) - metrics.stab_len, 0)
        plt.axvspan(it[start], it[-1], alpha=0.12)
    plt.xlabel("iteration")
    plt.ylabel("score")
    plt.legend()
    plt.title("VCLI-G / SCI timeline (iterative)")
    plt.tight_layout()
    plt.show()

def plot_hysteresis_ribbon(df: pd.DataFrame) -> None:
    it = _iter_axis(df)
    v  = _col(df, "VCLI_G")
    base = np.zeros_like(v)
    plt.figure(figsize=(8, 3.8))
    plt.plot(it, v)
    plt.fill_between(it, base, v, alpha=0.15)
    plt.xlabel("iteration")
    plt.ylabel("VCLI-G")
    plt.title("Hysteresis ribbon (VCLI-G)")
    plt.tight_layout()
    plt.show()

# ---- one-call runner (reads df, writes optional JSON, no mutations) ----
def run_sequence_metrics_addon(df: pd.DataFrame,
                               epsilon: float = 0.05,
                               write_json_path: Optional[str] = None,
                               append_ledger_path: Optional[str] = None,
                               do_plots: bool = True) -> SequenceMetrics:
    m = compute_sequence_metrics_from_df(df, epsilon=epsilon)
    if write_json_path:
        with open(write_json_path, "w", encoding="utf-8") as f:
            json.dump({"sequence_metrics": m.to_dict()}, f, indent=2)
    if append_ledger_path:
        try:
            with open(append_ledger_path, "r", encoding="utf-8") as f:
                data = json.load(f)
            if not isinstance(data, list):
                data = []
        except FileNotFoundError:
            data = []
        data.append({"sequence_metrics": m.to_dict()})
        with open(append_ledger_path, "w", encoding="utf-8") as f:
            json.dump(data, f, indent=2)
    if do_plots:
        plot_vcli_sci_timeline(df, m)
        plot_hysteresis_ribbon(df)
    return m
```

```python
# === Add-On: VCLI-G Geometry Panel (0910G-safe, no df edits) ===
# Reads canon df; computes geometry composites + minimal diagnostics.
# Advisory-only. Does NOT rename or mutate df.

import numpy as np, pandas as pd, json
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Optional, Dict

def _col(df: pd.DataFrame, *names: str, required=True, fill=np.nan):
    for n in names:
        if n in df.columns:
            return pd.to_numeric(df[n], errors="coerce").to_numpy()
    if required:
        raise KeyError(f"Missing required column: {' / '.join(names)}")
    return np.full(len(df), fill, dtype=float)

def _iter_axis(df: pd.DataFrame):
    if "iter" in df.columns: return pd.to_numeric(df["iter"], errors="coerce").to_numpy()
    if "sequence_idx" in df.columns: return pd.to_numeric(df["sequence_idx"], errors="coerce").to_numpy()
    return np.arange(len(df), dtype=float)

def _zscore(x: np.ndarray):
    m = np.nanmean(x); s = np.nanstd(x)
    if s == 0 or np.isnan(s): return np.zeros_like(x)
    return (x - m) / s

@dataclass
class GeoPanel:
    G1_focus: float    # avg z of focal instability (path_len + path_curv)
    G2_void: float     # avg z of void topology pressure
    G3_torque: float   # avg z of curvature/turning pressure
    G4_occ: float      # avg z of occlusion/orientation entropy
    corr_VCLI_G4: float  # Pearson r between VCLI_G and G4_H
    corr_VCLI_G2cut: float  # Pearson r between VCLI_G and G2_cut
    def to_dict(self) -> Dict:
        return {k: float(getattr(self,k)) for k in self.__dataclass_fields__.keys()}

def compute_geometry_panel(df: pd.DataFrame) -> GeoPanel:
    # pull base
    VCLI_G = _col(df, "VCLI_G")
    SCI    = _col(df, "SCI")

    # G1 (centroid wander / focus instability)
    g1_len = _col(df, "G1_path_len", "G1_L")
    g1_cur = _col(df, "G1_path_curv", "G1_K")
    z1 = np.nanmean(np.vstack([_zscore(g1_len), _zscore(g1_cur)]), axis=0)

    # G2 (void topology)
    g2_v   = _col(df, "G2_V")
    g2_chi = _col(df, "G2_chi")
    g2_ar  = _col(df, "G2_AR")
    g2_cut = _col(df, "G2_cut_depth", "G2_cut")
    z2 = np.nanmean(np.vstack([_zscore(g2_v), _zscore(g2_chi), _zscore(g2_ar), _zscore(g2_cut)]), axis=0)

    # G3 (curvature torque)
    g3_kv  = _col(df, "G3_k_var", "G3_kvar")
    g3_inf = _col(df, "G3_infl_density", "G3_infl")
    z3 = np.nanmean(np.vstack([_zscore(g3_kv), _zscore(g3_inf)]), axis=0)

    # G4 (orientation / occlusion entropy)
    g4_h   = _col(df, "G4_H")
    z4 = _zscore(g4_h)

    # Aggregate panel (means of per-row z)
    G1_focus = float(np.nanmean(z1))
    G2_void  = float(np.nanmean(z2))
    G3_torque= float(np.nanmean(z3))
    G4_occ   = float(np.nanmean(z4))

    # Simple correlations (nan-safe)
    def _corr(a,b):
        mask = ~(np.isnan(a) | np.isnan(b))
        if mask.sum() < 2: return np.nan
        return float(np.corrcoef(a[mask], b[mask])[0,1])

    corr_VCLI_G4    = _corr(VCLI_G, g4_h)
    corr_VCLI_G2cut = _corr(VCLI_G, g2_cut)

    return GeoPanel(G1_focus, G2_void, G3_torque, G4_occ, corr_VCLI_G4, corr_VCLI_G2cut)

def plot_geometry_timelines(df: pd.DataFrame) -> None:
    it = _iter_axis(df)
    # minimal, single-figure: VCLI_G with G4_H and G2_cut overlays for read
    VCLI_G = _col(df, "VCLI_G")
    g4_h   = _col(df, "G4_H")
    g2_cut = _col(df, "G2_cut_depth", "G2_cut")

    plt.figure(figsize=(8,4.8))
    plt.plot(it, VCLI_G, label="VCLI-G")
    plt.plot(it, g4_h,   label="G4_H (occlusion entropy)")
    plt.plot(it, g2_cut, label="G2_cut (void cut depth)")
    plt.xlabel("iteration / row index")
    plt.ylabel("value")
    plt.title("VCLI-G with geometry companions (G4_H, G2_cut)")
    plt.legend()
    plt.tight_layout()
    plt.show()

def plot_vclig_scatter(df: pd.DataFrame) -> None:
    # single scatter: VCLI_G vs G4_H (occlusion entropy)
    VCLI_G = _col(df, "VCLI_G")
    g4_h   = _col(df, "G4_H")
    plt.figure(figsize=(6.2,5.4))
    plt.scatter(g4_h, VCLI_G)
    plt.xlabel("G4_H (occlusion entropy)")
    plt.ylabel("VCLI-G (cognitive load)")
    plt.title("VCLI-G vs Occlusion Entropy")
    plt.tight_layout()
    plt.show()

def run_vclig_geometry_addon(df: pd.DataFrame,
                             write_json_path: Optional[str] = None) -> Dict:
    panel = compute_geometry_panel(df)
    plot_geometry_timelines(df)
    plot_vclig_scatter(df)
    out = {"vclig_geometry_panel": panel.to_dict()}
    if write_json_path:
        with open(write_json_path, "w", encoding="utf-8") as f:
            json.dump(out, f, indent=2)
    print(out)
    return out

# === USAGE ===
# out = run_vclig_geometry_addon(df, write_json_path="vclig_geometry_panel.json")
```

```python
# === Add-On: 2D Cognitive Space — VCLI-G × SCI (0910G-safe, no df edits) ===
# Reads canon df; does NOT rename or mutate it. Advisory-only.

import numpy as np, pandas as pd, json
import matplotlib.pyplot as plt
from typing import Dict

def _col(df: pd.DataFrame, name: str) -> np.ndarray:
    if name not in df.columns:
        raise KeyError(f"Required column missing: {name}")
    return pd.to_numeric(df[name], errors="coerce").to_numpy()

def _nanmask2(a: np.ndarray, b: np.ndarray):
    m = ~(np.isnan(a) | np.isnan(b))
    return a[m], b[m], m

def _convex_hull_area(x: np.ndarray, y: np.ndarray) -> float:
    """Monotone chain convex hull + polygon area. Returns 0 if <3 points."""
    pts = np.vstack([x, y]).T
    if pts.shape[0] < 3: return 0.0
    pts = pts[np.lexsort((pts[:,1], pts[:,0]))]

    def cross(o, a, b): return (a[0]-o[0])*(b[1]-o[1])-(a[1]-o[1])*(b[0]-o[0])
    lower=[]
    for p in pts:
        while len(lower)>=2 and cross(lower[-2], lower[-1], p) <= 0: lower.pop()
        lower.append(tuple(p))
    upper=[]
    for p in pts[::-1]:
        while len(upper)>=2 and cross(upper[-2], upper[-1], p) <= 0: upper.pop()
        upper.append(tuple(p))
    hull = np.array(lower[:-1] + upper[:-1], dtype=float)
    if hull.shape[0] < 3: return 0.0
    xh, yh = hull[:,0], hull[:,1]
    return 0.5*abs(np.dot(xh, np.roll(yh,-1)) - np.dot(yh, np.roll(xh,-1)))

def _principal_axis(x: np.ndarray, y: np.ndarray):
    """Eigen decomposition of covariance; returns eigvals (desc), eigvecs, and major-axis slope."""
    if x.size < 2:
        return (np.array([0.0,0.0]), np.eye(2), np.nan)
    X = np.vstack([x, y])
    C = np.cov(X)
    w, V = np.linalg.eig(C)
    idx = np.argsort(w)[::-1]
    w, V = w[idx], V[:, idx]
    # major eigenvector slope dy/dx
    vx = V[0,0]; vy = V[1,0]
    slope = np.nan if abs(vx) < 1e-12 else (vy / vx)
    return w, V, slope

def compute_space_summary(df: pd.DataFrame) -> Dict:
    V = _col(df, "VCLI_G")
    S = _col(df, "SCI")
    V, S, m = _nanmask2(V, S)

    if V.size == 0:
        return {
            "space_summary": {
                "n": 0, "centroid": {"VCLI_G": np.nan, "SCI": np.nan},
                "median": {"VCLI_G": np.nan, "SCI": np.nan},
                "hull_area": 0.0, "eigvals": [0.0, 0.0], "major_axis_slope": np.nan,
                "quadrant_counts": {"low_low":0,"low_high":0,"high_low":0,"high_high":0}
            }
        }

    # centroid & medians
    cV, cS = float(np.nanmean(V)), float(np.nanmean(S))
    mV, mS = float(np.nanmedian(V)), float(np.nanmedian(S))

    # hull area (spread in the cognitive plane)
    hull_area = float(_convex_hull_area(V, S))

    # principal axis (orientation of the cloud)
    eigvals, eigvecs, major_slope = _principal_axis(V, S)

    # quadrant counts using **medians** as the neutral split (robust to skew)
    q_ll = int(((V <  mV) & (S <  mS)).sum())
    q_lh = int(((V <  mV) & (S >= mS)).sum())
    q_hl = int(((V >= mV) & (S <  mS)).sum())
    q_hh = int(((V >= mV) & (S >= mS)).sum())

    return {
        "space_summary": {
            "n": int(V.size),
            "centroid": {"VCLI_G": cV, "SCI": cS},
            "median":   {"VCLI_G": mV, "SCI": mS},
            "hull_area": hull_area,
            "eigvals": [float(eigvals[0]), float(eigvals[1])],
            "major_axis_slope": float(major_slope) if major_slope==major_slope else None,
            "quadrant_counts": {
                "low_low": q_ll, "low_high": q_lh, "high_low": q_hl, "high_high": q_hh
            }
        }
    }

def plot_space(df: pd.DataFrame, summary: Dict, title: str = "VCLI-G × SCI (2D Cognitive Space)"):
    V = pd.to_numeric(df["VCLI_G"], errors="coerce").to_numpy()
    S = pd.to_numeric(df["SCI"], errors="coerce").to_numpy()
    mask = ~(np.isnan(V) | np.isnan(S))
    V, S = V[mask], S[mask]
    if V.size == 0:
        print("No valid points to plot."); return

    mV, mS = summary["space_summary"]["median"]["VCLI_G"], summary["space_summary"]["median"]["SCI"]
    cV, cS = summary["space_summary"]["centroid"]["VCLI_G"], summary["space_summary"]["centroid"]["SCI"]
    slope  = summary["space_summary"]["major_axis_slope"]

    plt.figure(figsize=(6.8, 5.6))
    plt.scatter(V, S)
    # medians (robust quadrant lines)
    plt.axvline(mV, linestyle="--", alpha=0.6)
    plt.axhline(mS, linestyle="--", alpha=0.6)
    # centroid
    plt.scatter([cV], [cS])
    plt.text(cV, cS, "centroid", ha="left", va="bottom")
    # major axis (principal orientation) through centroid
    if slope is not None:
        xs = np.array([V.min(), V.max()])
        ys = cS + slope*(xs - cV)
        plt.plot(xs, ys, linestyle=":")
    plt.xlabel("VCLI-G (cognitive load)")
    plt.ylabel("SCI (structural coherence)")
    plt.title(title)
    plt.tight_layout()
    plt.show()

def run_vclig_sci_space_addon(df: pd.DataFrame,
                              write_json_path: str = None,
                              title: str = "VCLI-G × SCI (2D Cognitive Space)") -> Dict:
    summary = compute_space_summary(df)
    plot_space(df, summary, title=title)
    if write_json_path:
        with open(write_json_path, "w", encoding="utf-8") as f:
            json.dump(summary, f, indent=2)
    print(summary)
    return summary

# === USAGE ===
# space = run_vclig_sci_space_addon(df, write_json_path="vclig_sci_space.json")
```

```python
# === Add-On: Cognitive Drift as a Curve (0910G-safe, no df edits) ===
# Reads canon df; computes ΔVCLI per iteration + stability/volatility metrics.
# Produces two single-figure charts and an optional JSON summary.

import numpy as np, pandas as pd, json
import matplotlib.pyplot as plt
from typing import Optional, Dict

def _series(df: pd.DataFrame, name: str) -> np.ndarray:
    if name not in df.columns:
        raise KeyError(f"Required column missing: {name}")
    return pd.to_numeric(df[name], errors="coerce").to_numpy()

def _iter_axis(df: pd.DataFrame):
    if "iter" in df.columns:
        return pd.to_numeric(df["iter"], errors="coerce").to_numpy()
    if "sequence_idx" in df.columns:
        return pd.to_numeric(df["sequence_idx"], errors="coerce").to_numpy()
    return np.arange(len(df), dtype=float)

def _first_sign_change(y: np.ndarray) -> Optional[int]:
    # index of first sign flip (ignoring zeros); returns None if no flip
    s = np.sign(y)
    # replace 0 with previous sign to avoid false breaks
    for i in range(1, s.size):
        if s[i] == 0: s[i] = s[i-1]
    idx = np.where(s[:-1] * s[1:] < 0)[0]
    return int(idx[0] + 1) if idx.size else None

def _trailing_stability(y: np.ndarray, eps: float) -> int:
    # length of trailing window where |y| < eps
    if y.size == 0: return 0
    run = 0
    for v in y[::-1]:
        if np.isnan(v) or abs(v) >= eps: break
        run += 1
    return int(run)

def compute_drift_metrics(df: pd.DataFrame, epsilon: float = 0.03) -> Dict:
    it = _iter_axis(df)
    v  = _series(df, "VCLI_G")

    # ΔVCLI per step (align with right edge so len matches it)
    dv = np.diff(v, prepend=v[:1])  # first element = 0 by definition

    # volatility & shape
    mean_drift   = float(np.nanmean(dv))
    std_drift    = float(np.nanstd(dv))
    max_up       = float(np.nanmax(dv)) if dv.size else 0.0
    max_down     = float(np.nanmin(dv)) if dv.size else 0.0
    max_accel    = float(np.nanmax(np.abs(dv))) if dv.size else 0.0
    cum_drift    = float(np.nansum(dv))  # equals v[-1]-v[0]

    # areas (positive/negative “work”)
    area_pos = float(np.nansum(np.clip(dv, 0, None)))
    area_neg = float(np.nansum(np.clip(dv, None, 0)))

    # inflection = first sign change (ignoring zeros)
    inflection_idx = _first_sign_change(dv)

    # stability length (trailing window where |ΔVCLI| < eps)
    stability_len = _trailing_stability(dv, eps=epsilon)

    # turning points count (sign flips)
    s = np.sign(dv)
    for i in range(1, s.size):
        if s[i] == 0: s[i] = s[i-1]
    turns = int(np.sum(s[:-1] * s[1:] < 0))

    out = {
        "drift_metrics": {
            "epsilon": float(epsilon),
            "mean_drift": mean_drift,
            "std_drift": std_drift,
            "max_increase": max_up,
            "max_decrease": max_down,
            "max_acceleration": max_accel,
            "cumulative_drift": cum_drift,
            "area_pos": area_pos,
            "area_neg": area_neg,
            "inflection_idx": None if inflection_idx is None else int(inflection_idx),
            "stability_len": stability_len,
            "turning_points": turns,
            "n": int(v.size)
        }
    }
    return out, it, v, dv

def plot_drift_curve(it: np.ndarray, dv: np.ndarray, epsilon: float) -> None:
    plt.figure(figsize=(8, 4.6))
    plt.plot(it, dv)
    plt.axhline(0.0, linestyle="--", alpha=0.5)
    plt.axhline(+epsilon, linestyle=":", alpha=0.5)
    plt.axhline(-epsilon, linestyle=":", alpha=0.5)
    plt.xlabel("iteration")
    plt.ylabel("ΔVCLI-G (change)")
    plt.title("Cognitive Drift Curve (ΔVCLI per step)")
    plt.tight_layout()
    plt.show()

def plot_cumulative_drift(it: np.ndarray, dv: np.ndarray) -> None:
    cd = np.cumsum(dv)
    plt.figure(figsize=(8, 4.2))
    plt.plot(it, cd)
    plt.axhline(0.0, linestyle="--", alpha=0.5)
    plt.xlabel("iteration")
    plt.ylabel("cumulative ΔVCLI-G")
    plt.title("Cumulative Drift (relative to first step)")
    plt.tight_layout()
    plt.show()

def run_drift_addon(df: pd.DataFrame,
                    epsilon: float = 0.03,
                    write_json_path: Optional[str] = None) -> Dict:
    summary, it, v, dv = compute_drift_metrics(df, epsilon=epsilon)
    plot_drift_curve(it, dv, epsilon=summary["drift_metrics"]["epsilon"])
    plot_cumulative_drift(it, dv)
    if write_json_path:
        with open(write_json_path, "w", encoding="utf-8") as f:
            json.dump(summary, f, indent=2)
    print(summary)
    return summary
```

Cumulative drift: tracks the net change in load from the starting image; flat tails imply stabilization, while slopes indicate continuing evolution.

```python
# ===============================================
# OVERLAY VIEW: HOW THE NOTEBOOK "SEES" YOUR IMAGE
# Original + gradient field + voids + skeleton
# ===============================================

import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.filters import sobel
from skimage.morphology import skeletonize
from skimage import exposure


def visualize_internal_overlays(path, figsize=(16, 10)):
    """
    Show how the VCLI/VTL-style pipeline "sees" an image:
      - Original
      - Original + gradient magnitude (pressure field)
      - Original + void mask (breathing zones)
      - Original + gradient skeleton (structural spine)
    """
    # ---- Load & preprocess ----
    img_bgr = cv2.imread(path)
    if img_bgr is None:
        raise ValueError(f"Could not load image: {path}")

    # Convert BGR (OpenCV) -> RGB (matplotlib)
    img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    gray_f = gray.astype("float32") / 255.0

    # ---- Sobel gradients ----
    sobel_x = cv2.Sobel(gray_f, cv2.CV_32F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(gray_f, cv2.CV_32F, 0, 1, ksize=3)

    # ---- Gradient magnitude (pressure field) ----
    grad_mag = np.sqrt(sobel_x**2 + sobel_y**2)
    grad_mag_norm = exposure.rescale_intensity(grad_mag, out_range=(0, 1))

    # ---- Void mask (low-pressure / breathing zones) ----
    void_thresh = np.percentile(grad_mag, 60)
    void_mask = grad_mag < void_thresh

    # ---- Skeleton of high-gradient regions ----
    skel_thresh = np.percentile(grad_mag_norm, 85)
    skeleton = skeletonize(grad_mag_norm > skel_thresh)

    # ---- Plot overlays ----
    fig, axes = plt.subplots(2, 2, figsize=figsize)
    ax = axes.ravel()

    # 1. Original
    ax[0].imshow(img)
    ax[0].set_title("Original Image")
    ax[0].axis("off")

    # 2. Gradient magnitude overlay (pressure field)
    ax[1].imshow(img)
    ax[1].imshow(grad_mag_norm, cmap="magma", alpha=0.5)
    ax[1].set_title("Original + Gradient Magnitude (Pressure Field)")
    ax[1].axis("off")

    # 3. Void mask overlay (breathing space)
    ax[2].imshow(img)
    void_overlay = np.zeros((*void_mask.shape, 4), dtype="float32")
    void_overlay[..., 2] = 1.0                             # blue channel
    void_overlay[..., 3] = void_mask.astype(float) * 0.35  # alpha
    ax[2].imshow(void_overlay)
    ax[2].set_title("Original + Void Mask (Breathing Zones)")
    ax[2].axis("off")

    # 4. Skeleton overlay (structural spine)
    ax[3].imshow(img)
    skel_overlay = np.zeros((*skeleton.shape, 4), dtype="float32")
    skel_overlay[..., :3] = 1.0                            # white
    skel_overlay[..., 3] = skeleton.astype(float) * 0.9
    ax[3].imshow(skel_overlay)
    ax[3].set_title("Original + Gradient Skeleton (Structural Spine)")
    ax[3].axis("off")

    plt.tight_layout()
    plt.show()
```
